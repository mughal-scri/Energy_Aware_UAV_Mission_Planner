# 09 — Tick-by-tick fleet simulation

The simulator is the flight controller for this toy world: it advances one tick at a time, charges movement energy, confirms targets, resolves traffic, and records events. This notebook runs a seeded three-drone mission and regenerates the battery and timeline figures.

```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))

from uavplanner.viz import write_figures_5

figures = write_figures_5(ROOT / 'results')
for path in figures:
    print(path.relative_to(ROOT))
```

## Read the outputs

- `fig_battery.png` shows each drone's remaining battery percentage per tick.
- `fig_timeline.png` marks takeoff, target confirmation, and landing events.
- `fig_minimax_nodes.png` compares total nodes over 20 seeded positions.

The three-drone figure scenario has separate radial routes from the base and no interceptor, so it isolates traffic, reporting, and energy use. The evasion case is in notebook 10.
