# 02 — BFS, DFS, UCS, Greedy

Four searches on the same map. BFS counts steps, DFS flood-fill marks reachability, UCS is cheapest energy so far, Greedy follows Euclidean distance. 


```python
from pathlib import Path
import sys

import numpy as np

ROOT = Path.cwd().resolve().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))

from uavplanner.config import Config
from uavplanner.informed import greedy
from uavplanner.models import EnergyCoef
from uavplanner.uninformed import bfs, dfs, ucs
from uavplanner.viz import fig_search_explored
from uavplanner.world import make_targets, make_world

cfg = Config(seed=42)
rng = np.random.default_rng(cfg.seed)
world = make_world(cfg, rng)
targets = make_targets(world, cfg, rng)
start = world.base
goal = next(t.pos for t in targets if not world.no_fly[t.pos])
coef = EnergyCoef()
panels = [
    ("BFS", bfs(world.no_fly, start, goal), start, goal),
    ("DFS", dfs(world.no_fly, start, goal), start, goal),
    ("UCS", ucs(world, world.no_fly, start, goal, coef, 1.0), start, goal),
    ("Greedy", greedy(world, world.no_fly, start, goal, coef, 1.0), start, goal),
]
for title, result, *_ in panels:
    print(f"{title:7} expanded={result.expanded:5} cost={result.cost:.3f} steps={max(len(result.path)-1, 0)}")
out = ROOT / "results" / "fig_search_explored.png"
fig_search_explored(world.no_fly, panels, out)
print("saved", out)
```
