# 08 — Mission planner

The planner is the desk before anyone takes off. It blocks restricted cells, rejects unreachable targets, assigns the rest with the CSP, then lets Hill Climbing and the genetic algorithm each propose a visiting order. The cheaper order is kept. If that tour still exceeds the usable battery, the energy allowance is tightened and the assignment is tried again. A storm is rejected before any of that search.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))

import numpy as np
from uavplanner.config import Config
from uavplanner.models import Drone, EnergyCoef
from uavplanner.planner import format_plan, plan_mission
from uavplanner.world import make_targets, make_world

cfg = Config(wind_preset='moderate', grid_n=30)
rng = np.random.default_rng(7)
world = make_world(cfg, rng)
targets = make_targets(world, cfg, rng)
drones = [
    Drone(id=i, battery=cfg.battery, capacity_kg=cfg.capacity_kg, pos=world.base)
    for i in (1, 2, 3)
]
plan = plan_mission(world, drones, targets, EnergyCoef(), cfg, np.random.default_rng(7))
print(format_plan(plan))
```

```python
storm_cfg = Config(wind_preset='storm', grid_n=30)
storm_rng = np.random.default_rng(7)
storm_world = make_world(storm_cfg, storm_rng)
storm_targets = make_targets(storm_world, storm_cfg, storm_rng)
storm_drones = [
    Drone(id=i, battery=storm_cfg.battery, capacity_kg=storm_cfg.capacity_kg, pos=storm_world.base)
    for i in (1, 2, 3)
]
storm = plan_mission(
    storm_world,
    storm_drones,
    storm_targets,
    EnergyCoef(),
    storm_cfg,
    np.random.default_rng(7),
)
print(format_plan(storm))
```
