# 05 — Genetic Algorithm for target order

The GA keeps a population of target orders. Tournament selection chooses parents, order crossover preserves a valid permutation, swap mutation adds variety, and elitism keeps the best route from being lost.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))

import numpy as np
from uavplanner.genetic import genetic_order
from uavplanner.hill_climbing import hill_climb
from uavplanner.viz import ordering_scaling_experiment, synthetic_ordering_matrix, write_figures_3

rng = np.random.default_rng(42)
cost, points = synthetic_ordering_matrix(12, rng)
stops = list(range(1, 13))
ga_order, ga_cost, ga_history = genetic_order(cost, stops, rng=rng)
hc_order, hc_cost, hc_history = hill_climb(cost, stops, rng=rng)
print(f'GA cost: {ga_cost:.3f}; Hill Climbing cost: {hc_cost:.3f}')
print(f'GA order is a permutation: {sorted(ga_order) == stops}')
```

```python
rows = ordering_scaling_experiment()
for row in rows:
    print({key: round(value, 3) for key, value in row.items()})

tour_png, convergence_png, scaling_png = write_figures_3(ROOT / 'results')
print(f'Wrote {tour_png}')
print(f'Wrote {convergence_png}')
print(f'Wrote {scaling_png}')
```
