# 04 — Hill Climbing for target order

A route matrix is directed: travelling A → B can cost more than B → A. Hill Climbing starts with an order, tries every swap and 2-opt reversal, and takes the best improving neighbour. Random restarts help it escape a poor initial order.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))

import numpy as np
from uavplanner.hill_climbing import brute_force_order, hill_climb, tour_cost
from uavplanner.viz import synthetic_ordering_matrix

rng = np.random.default_rng(42)
cost, points = synthetic_ordering_matrix(6, rng)
stops = list(range(1, 7))
baseline = tour_cost(stops, cost)
order, value, history = hill_climb(cost, stops, restarts=20, rng=rng)
exact_order, exact_value = brute_force_order(cost, stops)
print(f'Initial order: {stops}, cost={baseline:.3f}')
print(f'Hill Climbing: {order}, cost={value:.3f}')
print(f'Brute force:   {exact_order}, cost={exact_value:.3f}')
```

```python
import matplotlib.pyplot as plt

plt.plot(history, color='#009e73')
plt.title('Best Hill Climbing tour cost')
plt.xlabel('improvement step / restart')
plt.ylabel('tour energy (synthetic units)')
plt.grid(alpha=0.25)
plt.show()
```
